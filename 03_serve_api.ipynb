Call the `/shout` route, which returns a caption uppercased. Set `BASE` to the URL above.

```python
import json, urllib.request
BASE = 'https://...'  # <-- the service URL from `pxt service list`
body = json.dumps({'caption': 'hello from the api'}).encode()
req = urllib.request.Request(f'{BASE}/shout', data=body,
                             headers={'Content-Type': 'application/json'})
print(urllib.request.urlopen(req).read().decode())
```
