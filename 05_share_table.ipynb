# Share a table (optional)

A cloud-hosted table can be published and read by someone else. You publish it to a `pxt://`
address; a recipient replicates it and reads it, with no key needed for a public table.

```python
# Your hosted database. Set your org slug (see `pxt org list`).
# This MUST match the [[tool.pixeltable.database]] entry in pyproject.toml.
DB = 'pxt://your-org:champs'  # <-- edit this
print('targeting', DB)
```

`publish` makes a shareable copy; `push` sends later updates. On the other side, `replicate`
makes a read-only local copy and `pull` gets the latest.

```python
import pixeltable as pxt

org = DB.split('://')[1].split(':')[0]  # your org slug
postcards = pxt.get_table(f'{DB}/postcards')
pxt.publish(source=postcards, destination_uri=f'pxt://{org}/shared_postcards', access='public')
postcards.push()

local = pxt.replicate(remote_uri=f'pxt://{org}/shared_postcards', local_path='shared_postcards')
local.pull()
local.select(local.caption, local.width).collect()
```
