# Get set up on Pixeltable Cloud

Install Pixeltable, add your API key, and confirm your account. About 5 minutes.

Pixeltable is an OLTP database for multimodal AI. On the cloud your data lives in a database you
address as `pxt://<your-org>:<db>`.

## Install Pixeltable

With this repo cloned (for the notebooks, `app.py`, and sample data), sync it:

````python
!uv sync
````

Alternatively, install Pixeltable straight from PyPI instead:

````python
!pip install 'pixeltable[serve]'   # or: uv add 'pixeltable[serve]'
````

## Get access and an API key

Pixeltable Cloud is in closed beta. First get access, through your invite or the waitlist.
Once you are in:

1. Sign in to the dashboard and create an org.
2. Open **API Keys**, create one, and copy it. You see it once.

Then give the key to Pixeltable. Persistent, in `~/.pixeltable/config.toml`:

```toml
[pixeltable]
api_key = 'your-api-key'
```

Or per session: `export PIXELTABLE_API_KEY=your-api-key` (see `.env.example`). Details:
https://docs.pixeltable.com/platform/configuration. Restart the kernel after setting your key.

## Confirm your key and your org

`pxt config` prints the resolved configuration; your `pixeltable.api_key` should show as
`<redacted>`. `pxt org list` prints your org slug, which you use in database URIs.

````python
!pxt config
!pxt org list
````
