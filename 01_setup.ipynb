# Get set up on Pixeltable Cloud

Goal: install Pixeltable, add your API key, and confirm your cloud database is live. About 5 minutes.

Pixeltable is an OLTP database for multimodal AI. Your data lives in a database addressed as
`pxt://<your-org>:main`. Every org starts with a `main` database, so you do not create one here.
You just connect to it.

## Install

````python
!uv sync
````

## Get access and an API key

Pixeltable Cloud is in closed beta. First get access, through your invite or the waitlist.
Once you are in:

1. Sign in to the dashboard and create an org. It comes with a database named `main`.
2. Open **API Keys**, create one, and copy it. You see it once.

Then give the key to Pixeltable one of two ways.

**Recommended, and persistent.** Put it in your config file at `~/.pixeltable/config.toml`:

```toml
[pixeltable]
api_key = 'your-api-key'
```

This survives new terminals and kernel restarts. Details:
https://docs.pixeltable.com/platform/configuration

**Or, per session.** Export an environment variable before launching Jupyter, or use a `.env`
file (see `.env.example`):

```bash
export PIXELTABLE_API_KEY=your-api-key
```

Restart the kernel after setting your key.

## Confirm Pixeltable sees your key

`pxt config` prints the resolved configuration. Your `pixeltable.api_key` should show as
`<redacted>`, next to its source (your `config.toml` or the environment).

````python
!pxt config
````
